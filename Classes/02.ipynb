Last class, we made Python calculate things:

```python

```

Python happily gave us:

```

```

But we were only seeing the result ourselves.

If we want our program to show something to **everyone** who runs it, we use the `print()` function.
For example:

````python
print(3*2 + 6*4 - 2**2)
````

You can also show **texts** as output. Like this,  
`print("hello world!")` or, `print('hello world!')`  
Notice that **text** needs to be between single quotation mark `' '` or double quotation mark `" "`. Why 🤔?

Without quotation marks, Python thinks they are instructions.

But text is **not** an instruction—we simply want Python to display it.

That's why we wrap text in quotation marks.

You can also think of it this way:

> Python doesn't try to understand what's inside quotation marks—it simply prints it.

````python
print("Hello, Python!")
print('Hello, Python!')
print(2 + 2)
print("2 + 2")
````

But this one will cause an error.

````python
print(Hello)  # This line will cause an error
              # as there is no instruction named Hello
````

We can print **multiple values** at a time also. Just separate them by commas.

````python
print(1, 3.14, 22/7, 'This is a string.', 'Wonderful!')
````

We can join multiple texts together. This is called **Concatenation**.

````python
print(2 + 2)
print('2' + '2')
# 2 ar 2 koto hoy? amra 22 dekhi sekhane. :)
# - banite BAL Shamim Osman
print('I' + 'am' + 'Yu')
print('I', 'am', 'Yu')
# What are the differences between 4th and 5th line???
````

````python
print('2' + 2)
````

You can multiply a text several times.

````python
print('I am sorry for nothing.' * 5) # this is python only feature
````

It looks weird to print so many text in one line. We can use a clever technic named **newline character** (`'\n'`).
When a programming language sees it, it immediatly goes to next line and continues the rest of the text.

````python
print('I am sorry for nothing.\n' * 5)
````
