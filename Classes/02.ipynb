## Class 02: Show It (Printing Output) 🖨️
---

Last class, we made Python calculate things:

```python
2 + 2
```

Python happily gave us:

```
4
```

But we were just watching what the results will be. This will not be shown to other users. To show our results we can use `print()` function.  
For example:

````python
print(3*2 + 6*4 - 2**2)
````

You can also show **texts** as output. Like this,  
`print("hello world!")` or, `print('hello world!')`  
Notice that **text** needs to be between single qoutation mark `' '` or double quotation mark `" "`. Why 🤔?

Cuz without quotation mark, python thinks them as instructions. But texts are not instruction to python, we just want to show something we wish. That's why we wrap them in quote.

You can also say python **ignores** what is between quotation mark, just shows them. For example,

````python
print("Hello, Python!")
print('Hello, Python!')
print(2 + 2)
print("2 + 2")
````

But this one will cause an error.

````python
print(Hello, Python!)  # This line will cause an error
                       # as there is no instruction named Hello and Python
````

We can print **multiple values** at a time also. Just separate them by commas.

````python
print(1, 3.14, 22/7, 'This is a string.', 'Wonderful!')
````

***Remember that***, python is an **case sensitive** language. Python recognizes `print()`, `Print()`, `PRINT()`, `prInt()` ... as different things. You have to know which one is correct and which one is not.
