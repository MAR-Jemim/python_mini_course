### What to do now?

We can convert the type of input into anything. This is called Type Conversion/Casting.

At first check this out.

```python
num = '3.1415926535897932384626433832'
print(type(num), '\t', num)

num = float(num)
print(type(num), '', num)

num = int(num)
print(type(num), '\t', num)

num = bool(num)
print(type(num), '\t', num)
```

See that, we can convert if they are allowed.

There are several conversion function.
- int()
- float()
- str()
- bool()
- set()
- etc...

But some conversions are not allowed. Like `"3.1416"` [str type with radix point] » `3` [int type]

```python
print( int("3.1416") )
```
