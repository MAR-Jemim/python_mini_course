### Variable
Variable is used to **store** value under a **name** to use the value
later as much as needed.  
Python variables loo like this.
```python
# name = value
word = "Supercalifragilisticexpialidocious"
pi = 3.1415926535897932384626433832
mark = 76
grade = 'A'
grade_point = 3.50
cgpa = 23 / 7
passed = True
```
There are some rules for naming a variable:
- You can use `A-Z`, `a-z`, `0-9`, `_`.
- You can not start a name with **number**. ~~1st~~, ~~53~~
- You can not use **space**, or any other **symbols**.
- You can not use some reserved **keywords** like `print, if, pass, True,
while` etc.

For example: `roll_1, PASS, res, _res, Res, rES, very_long_name` etc.

````python
name = "Akram Hosen Khan Bahadur Chowdhury"
# Akram sir was really a good teacher to me 😢.
result = 2 * 3 ** 3 + 9

print(name)
print(result, result, result)
````

There are several types of data to be stored.
The **data types** are:


1.   **int** (Integer: Whole number)
2.   **float** (Float: Fractional number)
3.   **str** (String: Collection of letters, numbers and symbols)
4.   **bool** (Boolean: `True` or `False`)

you can check the type of any data.

````python
print(type(1))
print(type(1.0))
print(type('a'))
print(type(False))
````

BTW, we are using `=` sign and we don't even know it's name.
It is called **Assignment operator**. Yes, we learned the name in the first class.

`name = value`  
Here, `=` tells python to store `value` in a box named `name`. That's so simple.
