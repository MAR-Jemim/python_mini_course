### Variable

later as much as needed.

There are some rules for naming a variable:
- You can use `A-Z`, `a-z`, `0-9`, `_`.
- You can not start a name with **number**. ~~1st~~, ~~53~~
- You can not use **space**, or any other **symbols**.
- You can not use some reserved **keywords** like `print, if, pass, True,
while` etc.

BTW, we are using `=` sign and we don't even know it's name.
It is called **Assignment operator**. Yes, we learned the name in the first class.

`name = value`  
Here, `=` tells python to store `value` in a box named `name`. That's so simple.

> Do not read it like "~~NAME equal to VALUE~~". Instead, it is read like "name gets the value of value" or, "store value in name"

This sign can also do some creepy things 👻.
```python
choco = 4 + 6        # stores 10
choco = choco + 5    # adds 5 with previous choco value 10 and store 15
choco = choco - 9    # subtracts 9 from previous choco value and store 6
```
At first glance, this line may look strange.
`choco = choco + 5`
you might think,
> how can choco be equal to choco + 5?

But remember this is not equal sign. This is **assignment operator**. It tells python to calculate right side
and then store the value on the left side.

Look at the current value of choco (10).
1. Calculate 10 + 5 → 15.
2. Store 15 back into choco.
3. Now choco contains 15.

The same thing can be done shortly.

````python
choco = 4 + 6 # stores 10
choco += 5    # adds 5 with previous choco value 10 and store 15
choco -= 9    # subtracts 9 from previous choco value and store 6
print(choco)
````

These are called **shorthand assignment**.
There are several of these:
- ` =` assignment
- `+=` add and assign
- `-=` subtract and assign
- `*=` multiply and assign
- `/=` divide and assign
- `...` most of the operators can be used like this.
