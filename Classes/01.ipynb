### Operators
Every symbol that does something is called an **operator**.

`2 + 2 = 4`

Here, which one is **operand** and which one is **operator**?
And, we call it an **addition operation**.

Operators can be categorized in different ways.

These can be differentiated by how many operands they are taking:
- Unary operator: `+, -, not` etc
- Binary operator: `+, -, *, /, and, or` etc
- Ternary operator: `if ... else`

Or these can be separated by what they are doing:
- Mathematical operator: Does math. The usual suspects: `+`, `-`, `*`, `/`, `//`, `%`, `**`.
- Comparison operator: Compares values and answers with `True` or `False`. Examples: `==`, `!=`, `<`, `>`, `<=`, `>=`.
- Logical operator: Combines or flips boolean values. Examples: `and`, `or`, `not`.
- Assignment operator: Puts a value into a variable. Example: `=`. Its cousins like `+=`, `-=`, `*=`, etc., update and assign in one go.
- Bitwise operator: Works directly with bits. Examples: `&`, `|`, `^`, `~`, `<<`, `>>`. (Don't worry, we'll meet them later. 👀)
- Membership operator: Checks whether something is inside another thing. Examples: `in`, `not in`.
- Identity operator: Checks whether two names point to the exact same object. Examples: `is`, `is not`.

> 💡 **Operator** does the work. **Operand** is what the operator works on.

For example:

```python
10 + 5
```

- `+` → operator
- `10` and `5` → operands

Simple, right? Operators are the verbs of Python—they make things happen.
